# Minimal MCP call against dug-mcp

Opens an SSE session to the server, runs the MCP handshake, calls one tool, prints the raw result.
Payload is the `search_variables_by_name` example from the [dug-mcp README](https://github.com/bdc-assist/dug-mcp).

```python
import json

from mcp import ClientSession
from mcp.client.sse import sse_client

URL = "https://dug-mcp.apps.renci.org/sse"
TOOL, ARGS = "search_variables_by_name", {"search_term": "bmi"}

async with sse_client(URL) as (read, write):
    async with ClientSession(read, write) as session:
        await session.initialize()
        result = await session.call_tool(TOOL, ARGS)

print("raw CallToolResult:")
print(result.model_dump_json(indent=2, exclude_none=True))
print("\ndecoded text block:")
print(json.dumps(json.loads(result.content[0].text), indent=2))
```
